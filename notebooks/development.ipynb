```python
def merge_stmt(df_source, df_target, merge_columns):
    merge_columns = ' AND '.join([f"s.{c} = t.{c}" for c in merge_columns])

    query = f'''
    MERGE INTO {df_target} t
    USING {df_source} s
    ON {merge_columns}
    WHEN MATCHED THEN UPDATE SET *
    WHEN NOT MATCHED THEN INSERT *
    '''
    print(query)
```
